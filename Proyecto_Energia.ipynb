**Nombre:** Electricity Production Dataset

**Descripción:** Este conjunto de datos ofrece una visión general completa de los datos de producción eléctrica mundial que abarcan 48 países entre 2010 y 2023.
```python
import os
import json
import zipfile
import pandas as pd
from kaggle.api.kaggle_api_extended import KaggleApi

def extract_from_api():
    # Leer el archivo kaggle.json desde la ruta del proyecto
    with open('.kaggle/kaggle.json', 'r') as f:
        token = json.load(f)
    os.environ['KAGGLE_USERNAME'] = token['username']
    os.environ['KAGGLE_KEY'] = token['key']

    # Inicializar y autenticar la API de Kaggle
    api = KaggleApi()
    api.authenticate()

    # Descargar el archivo ZIP
    dataset_path = 'sazidthe1/global-electricity-production'
    zip_name = 'global-electricity-production.zip'
    extract_folder = 'tmp_kaggle_data'

    if not os.path.exists(zip_name):
        api.dataset_download_files(dataset_path, path='.', unzip=False)

    # Extraer el CSV del ZIP
    with zipfile.ZipFile(zip_name, 'r') as zip_ref:
        zip_ref.extractall(extract_folder)
        csv_file = [f for f in zip_ref.namelist() if f.endswith('.csv')][0]

    df = pd.read_csv(os.path.join(extract_folder, csv_file))
    return df

df2 = extract_from_api()
```

```python
import pandas as pd

# Rango de fechas
fecha_inicio = "1973-01-01"
fecha_fin = "2024-01-01"

# Generar rango de fechas
dim_fecha = pd.DataFrame({
    "fecha": pd.date_range(start=fecha_inicio, end=fecha_fin, freq="D")
})

# Extraer atributos temporales
dim_fecha["año"] = dim_fecha["fecha"].dt.year
dim_fecha["mes"] = dim_fecha["fecha"].dt.month
dim_fecha["día"] = dim_fecha["fecha"].dt.day
dim_fecha["nombre_mes"] = dim_fecha["fecha"].dt.month_name()
dim_fecha["nombre_día"] = dim_fecha["fecha"].dt.day_name()
dim_fecha["número_semana"] = dim_fecha["fecha"].dt.isocalendar().week
dim_fecha["trimestre"] = dim_fecha["fecha"].dt.quarter
dim_fecha["es_fin_de_semana"] = dim_fecha["nombre_día"].isin(["Saturday", "Sunday"])
dim_fecha["es_día_laboral"] = ~dim_fecha["es_fin_de_semana"]
```

### **Creamos una base datos relacional de MySQL en el servicio de : https://railway.app/ y nos conectamos a ella**

```python
from dotenv import load_dotenv
import os
from sqlalchemy import create_engine

def conectar_db():
    """
    Carga las variables de entorno y devuelve un engine de conexión SQLAlchemy a MySQL en Railway.
    """
    # Cargar variables del archivo .env
    load_dotenv()

    # Obtener las variables desde el entorno
    usuario = os.getenv("MYSQL_USER")
    contraseña = os.getenv("MYSQL_PASSWORD")
    host = os.getenv("MYSQL_HOST")
    puerto = os.getenv("MYSQL_PORT")
    base_de_datos = os.getenv("MYSQL_DB")

    # Validación rápida
    if not all([usuario, contraseña, host, puerto, base_de_datos]):
        raise ValueError("⚠️ Faltan variables de entorno en el archivo .env")

    # Crear y devolver el engine
    return create_engine(f"mysql+pymysql://{usuario}:{contraseña}@{host}:{puerto}/{base_de_datos}")
```

**Cargamos la tabla de fechas**

```python
#Creamos la tabla de fechas en la base de datos
engine = conectar_db()
dim_fecha.to_sql('dim_fecha', con=engine, if_exists='replace', index=False)
print("Tabla 'dim_fecha' creada en la base de datos MySQL.")
```

```python
# Cargar opciones únicas
def cargar_opciones():
    parametros = pd.read_sql("SELECT DISTINCT `Parámetro` FROM dim_electricidad", engine)['Parámetro'].dropna().tolist()
    paises = pd.read_sql("SELECT DISTINCT `Nombre del País` FROM dim_electricidad", engine)['Nombre del País'].dropna().tolist()
    años = pd.read_sql("SELECT DISTINCT YEAR(Fecha) AS Año FROM dim_electricidad", engine)['Año'].dropna().tolist()
    return sorted(parametros), sorted(paises), sorted(años)

parametros, paises, años = cargar_opciones()

# App Dash
app = dash.Dash(__name__)
server = app.server

app.layout = html.Div([
    html.H1("Consumo Energético por Trimestre", style={'textAlign': 'center'}),

    html.Div([
        html.Label("Año(s)"),
        dcc.Dropdown(id='filtro-año', options=[{'label': a, 'value': a} for a in años],
                     value=[años[0]], multi=True)
    ], style={'width': '30%', 'display': 'inline-block'}),

    html.Div([
        html.Label("País(es)"),
        dcc.Dropdown(id='filtro-pais', options=[{'label': p, 'value': p} for p in paises],
                     value=[paises[0]], multi=True)
    ], style={'width': '30%', 'display': 'inline-block'}),

    html.Div([
        html.Label("Parámetro(s)"),
        dcc.Dropdown(id='filtro-parametro', options=[{'label': p, 'value': p} for p in parametros],
                     value=[parametros[0]], multi=True)
    ], style={'width': '30%', 'display': 'inline-block'}),

    dcc.Graph(id='grafico-trimestre')
])

@app.callback(
    Output('grafico-trimestre', 'figure'),
    Input('filtro-año', 'value'),
    Input('filtro-pais', 'value'),
    Input('filtro-parametro', 'value')
)
def actualizar_grafico(años, paises, parametros):
    if not años or not paises or not parametros:
        return px.bar(title="Selecciona al menos una opción en cada filtro")

    años_str = ','.join(str(a) for a in años)
    paises_str = ','.join(f"'{p}'" for p in paises)
    parametros_str = ','.join(f"'{p}'" for p in parametros)

    query = f"""
    SELECT e.`Valor`, f.`trimestre`, f.`año`
    FROM dim_electricidad e
    JOIN dim_fecha f ON e.Fecha = f.fecha
    WHERE f.año IN ({años_str})
      AND e.`Nombre del País` IN ({paises_str})
      AND e.`Parámetro` IN ({parametros_str})
    """
    df = pd.read_sql(query, engine)

    if df.empty:
        return px.bar(title="No hay datos para los filtros seleccionados")

    df_grouped = df.groupby(["año", "trimestre"])["Valor"].sum().reset_index()

    fig = px.bar(
        df_grouped,
        x="trimestre",
        y="Valor",
        color="año",
        barmode="group",
        title="Consumo por Trimestre - Filtrado por Año, País y Parámetro",
        labels={"Valor": "Consumo", "trimestre": "Trimestre"}
    )

    return fig

if __name__ == '__main__':
    app.run(debug=True)
```

```python
import pandas as pd
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
from sqlalchemy import create_engine

# Cargar datos
def cargar_datos(tabla):
    if tabla == "dim_renovables":
        query = """
        SELECT `Año`, `Energía Hidroeléctrica`, `Energía Geotérmica`,
               `Energía Solar`, `Energía Eólica`, `Energía de la Madera`,
               `Energía de Residuos`, `Etanol Combustible (sin desnaturalizar)`,
               `Pérdidas y Coproductos de Biomasa`, `Energía de Biomasa`,
               `Energía Renovable Total`, `Diésel Renovable`, `Otros Biocombustibles`,
               `Energía Hidroeléctrica Convencional`, `Biodiésel`, `fecha`
        FROM dim_renovables
        """
        df = pd.read_sql(query, engine)
        return df.groupby("Año").sum(numeric_only=True).reset_index()
    else:
        query = """
        SELECT Fecha AS fecha, `Valor`, `Nombre del País`
        FROM dim_electricidad
        WHERE `Valor` > 0
        ORDER BY fecha
        """
        return pd.read_sql(query, engine)

# App Dash
app = dash.Dash(__name__)
server = app.server

app.layout = html.Div([
    html.H1("Dashboard de Energía - Data Warehouse", style={'textAlign': 'center'}),

    dcc.Dropdown(
        id='tabla-dropdown',
        options=[
            {'label': 'Consumo Renovable', 'value': 'dim_renovables'},
            {'label': 'Producción Eléctrica', 'value': 'dim_electricidad'}
        ],
        value='dim_renovables',
        clearable=False
    ),

    html.Div(id='contenedor-graficas')
])

@app.callback(
    Output('contenedor-graficas', 'children'),
    Input('tabla-dropdown', 'value')
)
def actualizar_graficas(tabla):
    try:
        df = cargar_datos(tabla)
        if df.empty:
            return html.Div("⚠️ No hay datos disponibles.")

        if tabla == "dim_renovables":
            columnas_energia = df.columns.drop("Año").tolist()
            df_melted = df.melt(id_vars='Año', value_vars=columnas_energia,
                                var_name='Tipo de Energía', value_name='Consumo')
            fig = px.line(df_melted, x='Año', y='Consumo', color='Tipo de Energía',
                          title='Consumo de Energías Renovables por Año',
                          markers=True,
                          color_discrete_sequence=px.colors.qualitative.Set3)
        else:
            df_grouped = df.groupby("Nombre del País")["Valor"].sum().reset_index()
            df_grouped = df_grouped.sort_values(by="Valor", ascending=False)
            fig = px.bar(df_grouped, x="Nombre del País", y="Valor",
                         title="Producción Eléctrica Total por País",
                         color="Nombre del País",
                         color_discrete_sequence=px.colors.qualitative.Set3)

        return dcc.Graph(figure=fig)

    except Exception as e:
        return html.Div(f"🚨 Error: {str(e)}")

if __name__ == '__main__':
    app.run(debug=True)
```
