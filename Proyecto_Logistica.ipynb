### **Fuente B : Api proporcionado por Kaggle**

**Nombre:** Electricity Production Dataset

**Descripción:** Este conjunto de datos ofrece una visión general completa de los datos de producción eléctrica mundial que abarcan 48 países entre 2010 y 2023. Proporciona información valiosa sobre el balance, la producción y el valor de la generación eléctrica en diferentes regiones durante el período especificado. Centrado en la producción eléctrica, este conjunto de datos resulta fundamental para analistas energéticos, responsables políticos e investigadores que buscan comprender las tendencias energéticas globales, identificar patrones y desarrollar estrategias para la producción y el consumo de energía sostenibles.

**URL:** https://www.kaggle.com/datasets/sazidthe1/global-electricity-production

```python
! pip install kaggle
```

```python
import os
import json
import zipfile
import pandas as pd
from kaggle.api.kaggle_api_extended import KaggleApi

def extract_from_api():
    # Leer el archivo kaggle.json desde la ruta del proyecto
    with open('.kaggle/kaggle.json', 'r') as f:
        token = json.load(f)
    os.environ['KAGGLE_USERNAME'] = token['username']
    os.environ['KAGGLE_KEY'] = token['key']

    # Inicializar y autenticar la API de Kaggle
    api = KaggleApi()
    api.authenticate()

    # Descargar el archivo ZIP
    dataset_path = 'sazidthe1/global-electricity-production'
    zip_name = 'global-electricity-production.zip'
    extract_folder = 'tmp_kaggle_data'

    if not os.path.exists(zip_name):
        api.dataset_download_files(dataset_path, path='.', unzip=False)

    # Extraer el CSV del ZIP
    with zipfile.ZipFile(zip_name, 'r') as zip_ref:
        zip_ref.extractall(extract_folder)
        csv_file = [f for f in zip_ref.namelist() if f.endswith('.csv')][0]

    df = pd.read_csv(os.path.join(extract_folder, csv_file))
    return df

df2 = extract_from_api()
```

```python
df2.rename(columns={
    "country_name": "Nombre del País",
    "date": "Fecha",
    "parameter": "Parámetro",
    "product": "Producto",
    "value" : "Valor",
    "unit": "Unidad"
}, inplace=True)

df2.head()
```
