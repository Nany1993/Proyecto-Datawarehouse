### **Fuente B : Api proporcionado por Kaggle**

**Nombre:** Electricity Production Dataset

**Descripción:** Este conjunto de datos ofrece una visión general completa de los datos de producción eléctrica mundial que abarcan 48 países entre 2010 y 2023. Proporciona información valiosa sobre el balance, la producción y el valor de la generación eléctrica en diferentes regiones durante el período especificado. Centrado en la producción eléctrica, este conjunto de datos resulta fundamental para analistas energéticos, responsables políticos e investigadores que buscan comprender las tendencias energéticas globales, identificar patrones y desarrollar estrategias para la producción y el consumo de energía sostenibles.

**URL:** https://www.kaggle.com/datasets/sazidthe1/global-electricity-production

```python
! pip install kaggle
```

```python
import os
import json
import zipfile
import pandas as pd
from kaggle.api.kaggle_api_extended import KaggleApi

def extract_from_api():
    # Leer el archivo kaggle.json desde la ruta del proyecto
    with open('.kaggle/kaggle.json', 'r') as f:
        token = json.load(f)
    os.environ['KAGGLE_USERNAME'] = token['username']
    os.environ['KAGGLE_KEY'] = token['key']

    # Inicializar y autenticar la API de Kaggle
    api = KaggleApi()
    api.authenticate()

    # Descargar el archivo ZIP
    dataset_path = 'sazidthe1/global-electricity-production'
    zip_name = 'global-electricity-production.zip'
    extract_folder = 'tmp_kaggle_data'

    if not os.path.exists(zip_name):
        api.dataset_download_files(dataset_path, path='.', unzip=False)

    # Extraer el CSV del ZIP
    with zipfile.ZipFile(zip_name, 'r') as zip_ref:
        zip_ref.extractall(extract_folder)
        csv_file = [f for f in zip_ref.namelist() if f.endswith('.csv')][0]

    df = pd.read_csv(os.path.join(extract_folder, csv_file))
    return df

df2 = extract_from_api()
```

```python
df2.rename(columns={
    "country_name": "Nombre del País",
    "date": "Fecha",
    "parameter": "Parámetro",
    "product": "Producto",
    "value" : "Valor",
    "unit": "Unidad"
}, inplace=True)

df2.head()
```

### **2.2. Creación de base de datos**

### **Vamos a crear una dimensión de fecha que me relacione ambas bases de datos**

```python
#vamos a indentificar las fecha maximay minima de cada dataframe
print("Fecha mínima en df1:", df1['fecha'].min())
print("Fecha máxima en df1:", df1['fecha'].max())
print("Fecha mínima en df2:", df2['Fecha'].min())
print("Fecha máxima en df2:", df2['Fecha'].max())
```

```python
import pandas as pd

# Rango de fechas
fecha_inicio = "1973-01-01"
fecha_fin = "2024-01-01"

# Generar rango de fechas
dim_fecha = pd.DataFrame({
    "fecha": pd.date_range(start=fecha_inicio, end=fecha_fin, freq="D")
})

# Extraer atributos temporales
dim_fecha["año"] = dim_fecha["fecha"].dt.year
dim_fecha["mes"] = dim_fecha["fecha"].dt.month
dim_fecha["día"] = dim_fecha["fecha"].dt.day
dim_fecha["nombre_mes"] = dim_fecha["fecha"].dt.month_name()
dim_fecha["nombre_día"] = dim_fecha["fecha"].dt.day_name()
dim_fecha["número_semana"] = dim_fecha["fecha"].dt.isocalendar().week
dim_fecha["trimestre"] = dim_fecha["fecha"].dt.quarter
dim_fecha["es_fin_de_semana"] = dim_fecha["nombre_día"].isin(["Saturday", "Sunday"])
dim_fecha["es_día_laboral"] = ~dim_fecha["es_fin_de_semana"]
```

```python
#Nos conectamos a la base de datos MySQL creadaa en railway
from dotenv import load_dotenv
import os
from sqlalchemy import create_engine

# Cargar las variables del archivo .env
load_dotenv()

# Obtener las variables desde el entorno
usuario = os.getenv("MYSQL_USER")
contraseña = os.getenv("MYSQL_PASSWORD")
host = os.getenv("MYSQL_HOST")
puerto = os.getenv("MYSQL_PORT")
base_de_datos = os.getenv("MYSQL_DB")

# Crear el engine para conectarte
engine = create_engine(f"mysql+pymysql://{usuario}:{contraseña}@{host}:{puerto}/{base_de_datos}")
```

**Cargamos la tabla de fechas**

```python
#Creamos la tabla de fechas en la base de datos
dim_fecha.to_sql('dim_fecha', con=engine, if_exists='replace', index=False)
print("Tabla 'dim_fecha' creada en la base de datos MySQL.")
```

```python
#validamos la creación de la tabla
query = "SELECT * FROM dim_fecha LIMIT 5;"
df_fecha = pd.read_sql(query, con=engine)
print(df_fecha.head())
```

**Cargamos la tabla de Renewable Energy Consumption (df1)**

```python
#Creamos la tabla df1 en la base de datos
df1.to_sql('dim_renovables', con=engine, if_exists='replace', index=False)
print("Tabla 'dim_renovables' creada en la base de datos MySQL.")
#validamos la creación de la tabla.
query = "SELECT * FROM dim_renovables LIMIT 5;"
df_renovables = pd.read_sql(query, con=engine)
print(df_renovables.head())
```

**Cargamos la tabla de Electricity Production Dataset**

```python
df2.to_sql('dim_electricidad', con=engine, if_exists='replace', index=False)
print("Tabla 'dim_electricidad' creada en la base de datos MySQL.")
#validamos la creación de la tabla.
query = "SELECT * FROM dim_electricidad LIMIT 5;"
df_electricity = pd.read_sql(query, con=engine)
```
